```python
df_clean = df[~((df["ClaimNb"] > 0) & (df["ClaimAmount"] == 0))].copy()
df_clean.shape
```

```python
max(df_clean["BonusMalus"])
```

```python
freq_moyenne = df_clean["ClaimNb"].sum() / df_clean["Exposure"].sum()
#freq_moyenne = (df_clean["ClaimNb"]/df_clean["Exposure"]).mean()
freq_moyenne
```

```python
import pandas as pd

bins = [18, 25, 35, 45, 55, 65, 100]
df_clean["age_group"] = pd.cut(df_clean["DrivAge"], bins=bins)

freq_by_age = df_clean.groupby("age_group", observed=True).apply(
    lambda g: g["ClaimNb"].sum() / g["Exposure"].sum()
)
freq_by_age
```

```python
bins_bm = [50,60,80,100,120,150,250]
df_clean["bm_group"] = pd.cut(df_clean["BonusMalus"], bins=bins_bm)

freq_by_bm = df_clean.groupby("bm_group", observed=True).apply(
    lambda g: g["ClaimNb"].sum() / g["Exposure"].sum()
)
freq_by_bm
```
